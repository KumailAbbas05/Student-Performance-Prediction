# Student Performance Prediction
This notebook demonstrates a simple end-to-end regression workflow using synthetic student data.

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
df = pd.read_csv("../data/student_performance_synthetic.csv")
df.head()
```

```python
df.info()
```

```python
X = df[["Study_Hours", "Attendance_Percentage", "Assignment_Score", "Sleep_Hours", "Previous_Score"]]
y = df["Final_Score"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
```

```python
model = LinearRegression()
model.fit(X_train, y_train)
predictions = model.predict(X_test)
```

```python
mae = mean_absolute_error(y_test, predictions)
mse = mean_squared_error(y_test, predictions)
r2 = r2_score(y_test, predictions)
print("MAE:", round(mae, 2))
print("MSE:", round(mse, 2))
print("R2 Score:", round(r2, 2))
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(y_test, predictions)
plt.xlabel("Actual Final Score")
plt.ylabel("Predicted Final Score")
plt.title("Actual vs Predicted Student Scores")
plt.grid(True)
plt.show()
```
